**ANALISIS CUSTOMER RETENTION AND LOYALITY**

```python
customer = """
  select
    customer_name,
    max(order_date) as last_transaction,
    count(distinct transaction_id) as freq_transaction,
    sum(after_discount) as volume
  from `Latihan.full_data_cleaned`
  where is_net = 1 and is_valid = 1
  group by 1
"""
```

```python
#mengunduh data customer untuk diolah
segmentation = pd_gbq.read_gbq(customer, project_id = project_id)
segmentation
```

```python
#menetapkan status di setiap metrik
segmentation['recency_status'] = pd.qcut(segmentation['last_transaction'], q = 2, labels = ['Lama','Baru'])
segmentation['freq_status'] = pd.qcut(segmentation['freq_transaction'], q = 2, labels = ['Rendah','Tinggi'])
segmentation['volume_status'] = pd.qcut(segmentation['volume'], q = 2, labels = ['Rendah', 'Tinggi'])
segmentation
```

```python
#menetapkan kondisi
condition = [
    (segmentation['recency_status'] == 'Baru') & (segmentation['freq_status'] == 'Tinggi') & (segmentation['volume_status'] == 'Tinggi'),
    (segmentation['recency_status'] == 'Baru') & (segmentation['freq_status'] == 'Tinggi') & (segmentation['volume_status'] == 'Rendah'),
    (segmentation['recency_status'] == 'Baru') & (segmentation['freq_status'] == 'Rendah') & (segmentation['volume_status'] == 'Tinggi'),
    (segmentation['recency_status'] == 'Baru') & (segmentation['freq_status'] == 'Rendah') & (segmentation['volume_status'] == 'Rendah'),
    (segmentation['recency_status'] == 'Lama') & (segmentation['freq_status'] == 'Tinggi') & (segmentation['volume_status'] == 'Tinggi'),
    (segmentation['recency_status'] == 'Lama' ) & (segmentation['freq_status'] == 'Tinggi') & (segmentation['volume_status'] == 'Rendah'),
    (segmentation['recency_status'] == 'Lama') & (segmentation['freq_status'] == 'Rendah') & (segmentation['volume_status'] == 'Tinggi'),
    (segmentation['recency_status'] == 'Lama') & (segmentation['freq_status'] == 'Rendah') & (segmentation['volume_status'] == 'Rendah')
]
```

```python
#membuat label
labels = [
    'Champions',
    'Loyal Small Buyers',
    'Potential loyalist',
    'Triallist',
    'At Risk',
    'Churning Customers',
    'Dormant Spenders',
    'Lost Customers'
]
```

```python
segmentation['RFM_status'] = np.select(condition, labels, default = 'No Status') #penggabungan kategori dan label
segmentation
```

```python
segmentation_summary = segmentation.groupby('RFM_status').agg({
    'customer_name': 'count',
    'freq_transaction': ['mean','max'],
    'volume': ['mean', 'max']
})
segmentation_summary
```

```python
pd_gbq.to_gbq(
    segmentation,
    'Latihan.segmentation',
    project_id,
    if_exists = 'replace'
)
```

```python
funneling = """
  with f as (
  select
    count(case when status = 'Click' then 1 end) as c,
    count(case when status = 'Viewed' then 1 end) as V,
    count(case when status = 'Add to Cart' then 1 end) as atc,
    count(case when status = 'Purchase' then 1 end) as p,
  from `Latihan.funnel_detail_cleaned`)
  select
    round((c+v+atc+p)/(c+v+atc+p)*100,2) as click,
    round(((v+atc+p)/(c+v+atc+p))*100,2) as Viewperc,
    round(((atc+p)/(v+atc+p))*100,2) as addtocartperc,
    round((p/(atc+p))*100,2) as purchperc
  from f
"""
```

```python
convertion_rate = pd_gbq.read_gbq(funneling, project_id = project_id)
transposed_convertion_rate = convertion_rate.T
transposed_convertion_rate.reset_index(inplace = True)
transposed_convertion_rate.columns = ['funnel_stage' ,'Percentage']
transposed_convertion_rate
```

```python
pd_gbq.to_gbq(
    transposed_convertion_rate,
    'Latihan.convertion_rate',
    project_id,
    if_exists = 'replace'
)
```
